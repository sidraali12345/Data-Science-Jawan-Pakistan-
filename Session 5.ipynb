```python
eng = int(input("Enter English Marks"))
isl = int(input("Enter Islamiyat Marks"))
urdu = int(input("Enter Urdu Marks"))
maths = int(input("Enter Maths Marks"))
comp = int(input("Enter Computer Marks"))

obtain_marks = eng+isl+urdu+maths+comp
print("Obtain Marks: ",obtain_marks)

total_marks = 500
print("Total Marks: ",total_marks)

perc = obtain_marks/total_marks*100
print("Percentage: ",perc)

if perc >= 80 and perc <=100:
    print("A-One Grade")
elif perc >= 70 and perc <80:
    print("A Grade")
elif perc >= 60 and perc <70:
    print("B Grade")
elif perc >= 50 and perc <60:
    print("C Grade")
elif perc >= 40 and perc <50:
    print("D Grade")
elif perc >= 33 and perc <40:
    print("E Grade")
elif perc >= 0 and perc <33:
    print("Fail")
else:
    print("Please recheck inserted value as your outpot is unappropriate")
```
