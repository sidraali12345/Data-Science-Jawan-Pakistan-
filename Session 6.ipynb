## loop
A loop allows a program to execute a set of instructions repeatedly.

Loops help in reducing the amount of code you need to write
- for
- while

## for loop:
for loop is used to iterate over a sequence (like a list, tuple, or range) and execute the block of code once for each item in the sequence.

## while loop:
while loop continues to execute as long as a certain condition is true. Once the condition becomes false, the loop stops.

# ***Nested For Loop***

```python
week_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
time_slots = ['9AM-10AM', '10AM-11AM', '11AM-12PM']
```

```python
for day in week_days:
    print(day)
```

```python
time_slots
```

```python
for day in week_days:
    print(day)
    for slot in time_slots:
        print(slot)
    print('***')
```

```python
for day in week_days:
    for slot in time_slots:
        print(day,slot)
    print('***'*10)
```

```python
cities = []
count = ''

while count != 'quit':
    city = input('enter a city')
    if city == 'quit':
        count = 'quit'
    else:
        cities.append(city)
 
```

## list Comprehension

```python
Ec_Store_Orders = [
    {"order_id": 1, "customer_name": "Ali", "total_amount": 120, "status": "Completed"},
    {"order_id": 2, "customer_name": "Raheel", "total_amount": 80, "status": "Pending"},
    {"order_id": 3, "customer_name": "Saleem", "total_amount": 150, "status": "Completed"},
    {"order_id": 4, "customer_name": "khushnoor", "total_amount": 95, "status": "Completed"},
    {"order_id": 5, "customer_name": "Alizy", "total_amount": 200, "status": "Pending"},
    {"order_id": 6, "customer_name": "javaid", "total_amount": 85, "status": "Completed"},
    {"order_id": 7, "customer_name": "Talha", "total_amount": 150, "status": "Pending"},
    {"order_id": 8, "customer_name": "Naveed", "total_amount": 200, "status": "Completed"},
]
```

#### Write a Python program to create a list of "order_ids" for all orders where the total_amount is greater than 100 and the order status is "Completed".

```python
for order in Ec_Store_Orders:
    print(order)
```

```python
for order in Ec_Store_Orders:
    print(order['total_amount'] > 100 and order['status'] == 'Completed')
```

```python
for order in Ec_Store_Orders:
    if order['total_amount'] > 100 and order['status'] == 'Completed':
        print(order)
```

```python
for order in Ec_Store_Orders:
    if order['total_amount'] > 100 and order['status'] == 'Completed':
        print(order['order_id'])
```

```python
order_id = []

for order in Ec_Store_Orders:
    if order['total_amount'] > 100 and order['status'] == 'Completed':
        order_id.append(order['order_id'])
```

```python
order_id
```

```python
customer_name = [order['order_id'] for order in Ec_Store_Orders if order['total_amount'] > 100 and order['status'] == 'Completed' ]
```

```python
customer_name
```

```python
customer_name = [order['customer_name'] for order in Ec_Store_Orders if order['total_amount'] > 100 and order['status'] == 'Completed' ]
```
