# Q1

```python
# Stop sending emails after 3 users (Using for loop and break)
users = ["Ali", "Sara", "John", "Zara", "Mike"]
count = 0
for i in users:
   if count==3:
       print("Limit reached. Stop sending emails")
       break
   else:
       print(f"Email sent to {i}") 
       count += 1
```

# Q2

```python
# Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] 
critical = 40
for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring")
        break
    else:
        print(f'Temperature: {temp}')   
```

# Q3

```python
# Calculate total price of items in a shopping cart (Using For loop)
cart = [100, 250, 50, 300]
total_price = 0
for price in cart:
    total_price += price
print(f"Total_price: {total_price}" )
```

# Q4

```python
#Skip customers with unpaid bills (Using For loop and continue)

customers = [ 
                {"name": "Ali", "paid": True},

                {"name": "Sara", "paid": False},

                {"name": "John", "paid": True} 
]
for name in customers:
    if name["paid"] == False:
        continue
    print("Access granted to:", name['name'])
```

# Q6

```python
# Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains=[]
for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)
print(domains)
```

# Q8

```python
# Create a multiplication table using a nested for loop.
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
for num in numbers:
    for multi in multipliers:
        print(f"{num} * {multi} = {num*multi}")
```

# Q9

```python
# Keep asking the user for a password until the correct password is entered.
#Use a while loop.
correct_password = "python123" 
password = input("Enter the password")
while password != correct_password:
    print("Incorrect password. Try again.") 
    password = input("Enter the password")
print("Access granted!")
```
