```python
class student:
    def __init__(self,school_name,location):
        self.schoolname = school_name
        self.loc = location 
        print("welcome")
        
    
    def greet(self,name,age):
        self.Name = name
        self.Age = age
        print(f"My name is {self.Name}, and my age is: {self.Age}")

    def sub(self,sub):
        return "We Are learning oops"
```
