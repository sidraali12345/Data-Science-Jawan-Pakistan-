# Q2

```python

#Age Group Classification
age = int(input("Enter your age"))

#"Child" if the age is between 0 and 12 (inclusive).
if age>=0 and age< 12:
    print("Child")
    
#"Teenager" if the age is between 13 and 19 (inclusive).
elif age>= 13 and age< 19:
    print("Teenager")
    
#"Adult" if the age is between 20 and 64 (inclusive).
elif age>= 20 and age <64:
    print("Adult")
    
#"Senior" if the age is 65 or older.
else:
    print("Senior")
```

# Q8

```python
# Add a new book in inventory using method
inventory = { 
              "The Great Gatsby": {"price": 10.99, "quantity": 5}, 
              "Rich dad poor dad": {"price": 8.99, "quantity": 8}, 
              "Harry Potter": {"price": 12.50, "quantity": 2}, 
            }
inventory.update({"The Prisoner of Zenda": {"price": 21.90, "quantity": 6}})
print(inventory)

#get all book titles using method
Book_titles = list(inventory.keys())
print(Book_titles)

#Remove a book if it's out of stock

if inventory["Harry Potter"]["quantity"] == 0:
    inventory.pop("Harry Potter")
    print("out of stock")
else:
    print("It is in stock")
print(inventory)
```

```python
#Write a Python program to check whether the entered letter is a vowel, a consonant, or invalid input.
Alphabets = input("Enter a letter: ") 
if len(Alphabets) ==1 and Alphabets.isalpha():
    if Alphabets.lower() in ['a','e','i','o','u']:
        print("It is a vowel")
    else :
        print("It is a consonant")
else :
    print("Please enter valid Letter")
```
