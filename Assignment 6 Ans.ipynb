# Q1

```python
#write a Python program that: Has a balance of 1000.Asks the user for a withdrawal amount.Checks for these errors:
balance = 1000
try:
    amount = float(input("Enter amount to withdraw:"))
    if amount <=0:
        print("Amount must be more than zero.")
    elif amount > balance:
        print("Insufficient balance.")
    else:
        balance = balance - amount
        print("Remaining Amount:" , balance)

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Thank you for using the ATM.")
```

# Q7

```python
#Define a class named Book with the following:A constructor (init) that initializes three attributes: title, author, and year.
#A method display_info() that prints the book details in the format:
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def display_info(self):
        print(f"{self.title}  by {self.author}, published in {self.year}.")

    def is_classic(self):
        print("This is a book.")

my_book = Book("To Kill a Mockingbird", "Harper Lee", 1960)

my_book.display_info()
my_book.is_classic()
```

# Q8

```python
#Define a class named Student that has the following:Attributes: name, grade, and age.
class Student:
    def __init__(self, name, grade, age):
        self.name = name
        self.grade = grade
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")

    def show_grade(self):
        print(f"My grade is {self.grade}.")

student1 = Student("Alice", "A", 16)

student1.introduce()
student1.show_grade()
```

```python
# Write a program that:Asks the user to enter a parking slot number they want to reserve.Checks the slot number:
available_slots = [101, 102, 103, 104, 105]
reserved = [106, 107, 108, 109, 110]

try:
    slot = int(input("Enter the parking slot number you want to reserve: "))

    if slot in available_slots:
        print(f"Slot {slot} is reserved for you.")
    elif slot in reserved:
        print(f"Slot {slot} is already reserved by someone else.")
    else:
        print(f"Slot {slot} does not exist.")

except ValueError:
    print("Invalid input. Please enter a valid slot number.")
```
