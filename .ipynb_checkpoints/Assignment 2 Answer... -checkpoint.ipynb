# Q2

```python
# You are given a list that contains some duplicate items. Remove the duplicates by converting the list to a set.
shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]
Unique_shopping_list = set(shopping_list)
print(Unique_shopping_list)

# Final output should be in list
New_shopping_list =  list(Unique_shopping_list)
print(New_shopping_list)
```

# Q3

```python
#Using set methods
#Find students who are in Course A but not in Course B (difference)
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}
Difference= course_a.difference(course_b)
print(Difference)

#Find students who are only in one of the two courses (symmetric difference)
Symmetric_Difference= course_a.symmetric_difference(course_b)
print(Symmetric_Difference)
```

# Q4

```python
#Display the first 3 elements.
numbers = (10, 20, 30, 40, 50, 60, 70)
first_three_elements= numbers[:3]
print(first_three_elements)

# Display the last 3 elements
last_3_element= numbers[-3:]
print(last_3_element)

#Display alternate elements.
alternate_element=numbers[::2]
print(alternate_element)
```

# Q6

```python
#Set Comparison  Find common elements.
A = {1, 2, 3, 4, 5} 
B = {4, 5, 6, 7}
Common_Elements =A.intersection(B)
print(Common_Elements)

#Find elements only in Set A.
Elements_Only_SetA = A.difference(B)
print(Elements_Only_SetA)

# Find elements only in Set B
Elements_Only_SetB = B.difference(A)
print(Elements_Only_SetB)
```
