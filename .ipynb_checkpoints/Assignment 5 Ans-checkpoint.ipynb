# Q1

```python
# Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.
def count_vowels(text):
    vowels = "a,e,i,o,u,A,E,I,O,U"
    count= 0
    for char in text:
        if char in vowels:
            count +=1

    return f"{count} vowels"
print (count_vowels("Welcome"))
    
```

# Q6

```python
#Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments.
def calculate_rectangle(length,width):
    def calculate_area():
        return length * width
        
    def calculate_perimeter():
        return 2*(length * width)
        
    print("Area:" , calculate_area())
    print("Perimeter:", calculate_perimeter())
    
calculate_rectangle(5,3)
```

# Q8

```python
# Write a nested function to calculate the total hotel bill based on the number of nights.
def calculate_total_bill(nights):
    
    per_night_charges = 5000
    
    def room_cost():
        return nights * per_night_charges
        
    def tax():
        return room_cost() * 0.10
        
    def total():
        return tax()+ room_cost()
        
    print(f"Room Cost: {room_cost()}")
    print(f"Tax: {tax()}")
    print(f"Total_Bill : {total()}")
    
calculate_total_bill(3)
```

```python
# User Login Validation A website checks a user's username and password before allowing access.
def login_validation():
    username = input("Enter your username: ")
    password = input("Enter your password: ")
    def check_username():
        return username == "admin"
    def check_password():
        return password =="Admin123"
    if check_username() and check_password():
        print("Login Successfull")
    else:
        print("Invalid Username and Password")
    
login_validation()        
```

# Q10

```python
# Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:
def shopping_bill(price, quantity, discount=0):
    
    def calculate_total():
        return price * quantity
    def apply_discount(total):
        final_amount = total - discount
        return max(0, final_amount)
    total_price = calculate_total()
    final_amount = apply_discount(total_price)
    return final_amount
       
print(f" final bill= {shopping_bill(200,3)}")
print(f" final bill= {shopping_bill(200,3,100)}")
```
