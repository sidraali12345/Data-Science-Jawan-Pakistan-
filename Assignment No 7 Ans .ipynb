```python
#Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, 
# year of joining, education, and his year of experience.

class Employee:
    def __init__ (self,name, number, address, salary, post, year_of_joining, education,year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary= salary
        self.post= post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

#Make a method called describe_employee() that prints the detail Information about employee
    
    def describe_employee(self):
        print("\n ----Employee Detail Information-----")
        print("Name:",self.name)
        print("Employee Number:",self.number)
        print("Address:",self.address)
        print("Salary:",self.salary)
        print("Post:", self.post)
        print("Year Of Joining:",self.year_of_joining)
        print("Education:",self.education)
        print("Year OF Experience:",self.year_of_experience)

# method update_ year_of_experience()

    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print("Experience update successfully")

# method called increase_salary() that do increamnet in it's salary.


    def increase_salary(self, increment):
        self.salary += increment
        print("Salary Increased Successfully!")

#Make an instance called employee101 from your class.

employee101= Employee("sidra",101,"karachi",26000,"Data Analyst", 2022, "BS Computer Science", 4 )

# Printing four attributes individually
print("\n------- Four Attribute Individually----------")
print("Name:", employee101.name)
print("Employee Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)

# call all it's methods for describing employee, ssalary increamnet and increase year of experience.
employee101.describe_employee()
employee101.increase_salary(10000)
employee101.update_year_of_experience(6)

# Creating 10 different employee instances

employee1 = Employee("Ahmed", 102, "Karachi", 45000, "Developer", 2021, "BS IT", 4)
employee2 = Employee("Sara", 103, "Lahore", 55000, "Manager", 2020, "MBA", 5)
employee3 = Employee("Usman", 104, "Islamabad", 40000, "Designer", 2023, "BS Design", 2)
employee4 = Employee("Ayesha", 105, "Karachi", 60000, "Data Scientist", 2019, "MS Data Science", 6)
employee5 = Employee("Hassan", 106, "Lahore", 35000, "Accountant", 2024, "B.Com", 1)
employee6 = Employee("Fatima", 107, "Multan", 50000, "HR Officer", 2022, "MBA", 3)
employee7 = Employee("Bilal", 108, "Quetta", 42000, "Web Developer", 2021, "BS CS", 4)
employee8 = Employee("Zainab", 109, "Peshawar", 65000, "Project Manager", 2018, "MS IT", 7)
employee9 = Employee("Hamza", 110, "Hyderabad", 38000, "Marketing Officer", 2023, "BBA", 2)
employee10 = Employee("Noor", 111, "Karachi", 70000, "Software Engineer", 2017, "MS CS", 8)

employees = [employee1,employee2,employee3,employee4,employee5,employee6,employee7,employee8,employee9,employee10]
print("    \n\n--------- DESCRIPTIONS FOR 10 EMPLOYEES ----------")
#call describe_employee() for each instance
for emp in employees:
    emp.describe_employee()

# Updating some employees
print("\n")
employee1.increase_salary(5000)
employee2.update_year_of_experience(6)
employee3.increase_salary(8000)
employee4.update_year_of_experience(7)
print("\n")
    
    
```
